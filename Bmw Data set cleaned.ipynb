```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df.drop_duplicates()
print(df)
```

```python
df
```

```python
df["model"].unique()
```

```python
df["price"].mean()
```

```python
df["price"] = df["price"].astype("int64")
df["mileage"] = df["mileage"].astype("int64")
df.dtypes
```

```python
df.loc[df["price"].idxmax()]
```

```python
df.loc[df["price"].idxmin()]
```

```python
df
```

```python
df["cost_price"] = df["price"]*0.85 
df["profit"]= df["price"] - df["cost_price"]
```

```python
df
```

```python
df["Profit_Status"] = df["profit"].apply(
    lambda x: "High Profit" if x > 3000
    else ("Medium Profit" if x > 1500
    else ("Low Profit" if x > 0 else "Loss"))
)
```

```python
df
```

```python
# Price Distribution
plt.figure(figsize=(8,4))
sns.histplot(df["price"], bins=30)
plt.title("BMW Price Distribution")
plt.show()
```

```python
# price vs mileage
plt.figure(figsize=(8,4))
sns.scatterplot(data=df, x="mileage", y="price")
plt.title("Price vs Mileage")
plt.show()
```

```python
# Average price per model
plt.figure(figsize=(10,5))
sns.barplot(data=df, x="model", y="price", estimator=np.mean)
plt.xticks(rotation=45)
plt.title("Average Price vs BMW Model")
plt.show()
```

```python
df.groupby("model")["price"].mean().sort_values(ascending=False).head(1)
```

```python
# visualization
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x="year", y="price")
plt.title("BMW Price Trend by Year")
plt.show()
```
